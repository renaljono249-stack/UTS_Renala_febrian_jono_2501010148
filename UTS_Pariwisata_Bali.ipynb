```python

# ==========================================
# MEMBERSIHKAN DATASET BPS
# ==========================================

import pandas as pd
import os

# Membaca kembali file mentah tanpa mengubahnya
df_raw = pd.read_csv(file_raw, header=None)

# Daftar nama bulan, termasuk variasi penulisan dari BPS
nama_bulan = {
    "januari": "Januari",
    "pebruari": "Februari",
    "februari": "Februari",
    "maret": "Maret",
    "april": "April",
    "mei": "Mei",
    "juni": "Juni",
    "juli": "Juli",
    "agustus": "Agustus",
    "september": "September",
    "oktober": "Oktober",
    "nopember": "November",
    "november": "November",
    "desember": "Desember"
}

# Posisi baris header tahun dalam file CSV
baris_header = [3, 19, 35]

hasil = []

# Memproses setiap kelompok tahun
for i, posisi in enumerate(baris_header):

    # Mengambil tahun dari header
    tahun_header = df_raw.iloc[posisi, 1:10]

    # Menentukan batas akhir kelompok
    if i < len(baris_header) - 1:
        batas = baris_header[i + 1]
    else:
        batas = len(df_raw)

    # Membaca setiap baris bulan
    for r in range(posisi + 1, batas):

        bulan_asli = str(df_raw.iloc[r, 0]).strip().lower()

        # Hanya mengambil baris nama bulan
        if bulan_asli not in nama_bulan:
            continue

        bulan = nama_bulan[bulan_asli]

        # Membaca nilai untuk masing-masing tahun
        for kolom, tahun in enumerate(tahun_header, start=1):

            try:
                tahun = int(tahun)
            except (ValueError, TypeError):
                continue

            # Mengambil hanya tahun 2004 sampai 2025
            if tahun < 2004 or tahun > 2025:
                continue

            nilai = df_raw.iloc[r, kolom]

            # Menghapus spasi pemisah ribuan
            if pd.isna(nilai):
                continue

            nilai = str(nilai).replace(" ", "").replace(",", "")

            if not nilai.isdigit():
                continue

            hasil.append({
                "bulan": bulan,
                "tahun": tahun,
                "jumlah_wisatawan": int(nilai)
            })

# Membuat DataFrame hasil bersih
df_bps_bersih = pd.DataFrame(hasil)

# Mengurutkan data berdasarkan tahun dan bulan
urutan_bulan = list(nama_bulan.values())
urutan_bulan = list(dict.fromkeys(urutan_bulan))

df_bps_bersih["urutan_bulan"] = (
    df_bps_bersih["bulan"].map(
        {bulan: i for i, bulan in enumerate(urutan_bulan, start=1)}
    )
)

df_bps_bersih = df_bps_bersih.sort_values(
    ["tahun", "urutan_bulan"]
).drop(columns=["urutan_bulan"]).reset_index(drop=True)

# Menampilkan hasil
print("JUMLAH RECORD VALID:", len(df_bps_bersih))
print("JUMLAH KOLOM:", len(df_bps_bersih.columns))

print("\nTIPE DATA:")
print(df_bps_bersih.dtypes)

print("\nLIMA DATA PERTAMA:")
display(df_bps_bersih.head())

print("\nLIMA DATA TERAKHIR:")
display(df_bps_bersih.tail())

# Menyimpan hasil bersih di folder processed
folder_processed = "/content/proyek_uts_pariwisata/data/processed"
os.makedirs(folder_processed, exist_ok=True)

df_bps_bersih.to_csv(
    os.path.join(folder_processed, "wisatawan_domestik_bersih.csv"),
    index=False
)

print("\nFile data bersih berhasil disimpan!")
```

```python

# ==========================================
# TAHAP 11: AKUISISI DATA MENGGUNAKAN API
# Sumber: Open-Meteo Historical Weather API
# ==========================================

import requests
import pandas as pd
import os
import json
import time
import hashlib
from datetime import datetime
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
from getpass import getpass

# ------------------------------------------
# 1. MEMASUKKAN EMAIL UNTUK USER-AGENT
# ------------------------------------------

# Email dimasukkan saat runtime, bukan ditulis
# langsung di dalam kode notebook.
email_kampus = getpass(
    "Masukkan email kampus pribadi Anda: "
)

HEADERS = {
    "User-Agent": f"UTS-PariwisataBali/1.0 ({email_kampus})"
}

# ------------------------------------------
# 2. MENENTUKAN ENDPOINT DAN PARAMETER API
# ------------------------------------------

url_api = "https://archive-api.open-meteo.com/v1/archive"

params_api = {
    "latitude": -8.65,
    "longitude": 115.2167,
    "start_date": "2019-01-01",
    "end_date": "2025-12-31",
    "daily": [
        "weather_code",
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "precipitation_sum",
        "relative_humidity_2m_mean"
    ],
    "timezone": "Asia/Makassar"
}

# ------------------------------------------
# 3. MENYIAPKAN RETRY JIKA REQUEST GAGAL
# ------------------------------------------

session = requests.Session()

retry_api = Retry(
    total=5,
    connect=5,
    read=5,
    backoff_factor=1,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"]
)

adapter = HTTPAdapter(max_retries=retry_api)

session.mount("https://", adapter)

# ------------------------------------------
# 4. MENGIRIM REQUEST KE API
# ------------------------------------------

try:
    response_api = session.get(
        url_api,
        params=params_api,
        headers=HEADERS,
        timeout=(10, 60)
    )

    # Jeda setelah request
    time.sleep(2)

    print("Status HTTP:", response_api.status_code)

    response_api.raise_for_status()

    data_api = response_api.json()

    print("Request API berhasil!")

except requests.exceptions.RequestException as error:
    print("Terjadi kesalahan:", error)
    raise

# ------------------------------------------
# 5. MENYIMPAN JSON MENTAH TANPA DIUBAH
# ------------------------------------------

folder_raw = (
    "/content/proyek_uts_pariwisata/data/raw"
)

os.makedirs(folder_raw, exist_ok=True)

path_json = os.path.join(
    folder_raw,
    "cuaca_denpasar_raw.json"
)

# Menyimpan byte respons asli dari server
with open(path_json, "wb") as file:
    file.write(response_api.content)

print("JSON mentah berhasil disimpan!")

# ------------------------------------------
# 6. MENGHITUNG HASH MD5 JSON MENTAH
# ------------------------------------------

md5_api = hashlib.md5()

with open(path_json, "rb") as file:
    for bagian in iter(
        lambda: file.read(4096), b""
    ):
        md5_api.update(bagian)

hash_api = md5_api.hexdigest()

print("Hash MD5 API:", hash_api)

# ------------------------------------------
# 7. NORMALISASI JSON MENJADI DATAFRAME
# ------------------------------------------

daily = data_api["daily"]

df_api = pd.DataFrame({
    "tanggal": daily["time"],
    "kode_cuaca": daily["weather_code"],
    "suhu_rata_rata_c": daily["temperature_2m_mean"],
    "suhu_maksimum_c": daily["temperature_2m_max"],
    "suhu_minimum_c": daily["temperature_2m_min"],
    "curah_hujan_mm": daily["precipitation_sum"],
    "kelembapan_rata_rata_pct": daily["relative_humidity_2m_mean"]
})

df_api["tanggal"] = pd.to_datetime(
    df_api["tanggal"]
)

# ------------------------------------------
# 8. MEMERIKSA HASIL AKUISISI
# ------------------------------------------

print("JUMLAH RECORD API:", len(df_api))
print("JUMLAH KOLOM:", len(df_api.columns))

print("\nTIPE DATA:")
print(df_api.dtypes)

print("\nLIMA DATA PERTAMA:")
display(df_api.head())

print("\nINFORMASI DATA:")
df_api.info()

# ------------------------------------------
# 9. MENYIMPAN CSV HASIL NORMALISASI
# ------------------------------------------

folder_processed = (
    "/content/proyek_uts_pariwisata/data/processed"
)

os.makedirs(folder_processed, exist_ok=True)

path_csv_api = os.path.join(
    folder_processed,
    "cuaca_denpasar.csv"
)

df_api.to_csv(path_csv_api, index=False)

print("CSV hasil normalisasi berhasil disimpan!")

# ------------------------------------------
# 10. MEMBUAT PROVENANCE API
# ------------------------------------------

provenance_api = {
    "nama_dataset": "Cuaca Harian Denpasar",
    "sumber": "Open-Meteo Historical Weather API",
    "endpoint": url_api,
    "parameter": params_api,
    "tanggal_akses": datetime.now().astimezone().isoformat(),
    "metode": "GET",
    "status_http": response_api.status_code,
    "format_mentah": "JSON",
    "format_olahan": "CSV",
    "lisensi": "CC BY 4.0",
    "url_lisensi": "https://creativecommons.org/licenses/by/4.0/",
    "hash_md5": hash_api,
    "jumlah_record": len(df_api),
    "periode": "2019-01-01 sampai 2025-12-31",
    "user_agent": "UTS-PariwisataBali/1.0 (email kampus dimasukkan saat runtime)"
}

folder_provenance = (
    "/content/proyek_uts_pariwisata/provenance"
)

os.makedirs(folder_provenance, exist_ok=True)

with open(
    os.path.join(folder_provenance, "provenance_api.json"),
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        provenance_api,
        file,
        ensure_ascii=False,
        indent=4
    )

print("Provenance API berhasil disimpan!")
```

```python

# ==========================================
# TAHAP 12: PEMERIKSAAN ROBOTS.TXT
# ==========================================

import requests
import time
from urllib.parse import urlparse
from urllib.robotparser import RobotFileParser

# Halaman target scraping
url_scraping = "https://visitbali.com/best-attractions"

# Membuat alamat robots.txt
parsed_url = urlparse(url_scraping)

robots_url = (
    f"{parsed_url.scheme}://{parsed_url.netloc}/robots.txt"
)

# Menggunakan User-Agent dari tahap API
headers_scraping = {
    "User-Agent": HEADERS["User-Agent"]
}

# Mengambil robots.txt dengan timeout
response_robots = requests.get(
    robots_url,
    headers=headers_scraping,
    timeout=20
)

time.sleep(2)

print("URL robots.txt:", robots_url)
print("Status HTTP:", response_robots.status_code)

if response_robots.status_code == 200:

    # Membaca aturan robots.txt
    robot_parser = RobotFileParser()
    robot_parser.parse(response_robots.text.splitlines())

    # Memeriksa apakah halaman target boleh diakses
    boleh_scraping = robot_parser.can_fetch(
        headers_scraping["User-Agent"],
        url_scraping
    )

    print("Boleh scraping:", boleh_scraping)

    if not boleh_scraping:
        print("STOP: Halaman tidak diizinkan untuk scraping.")

elif response_robots.status_code == 404:
    print("robots.txt tidak ditemukan.")
    print("Jangan langsung menganggap scraping diizinkan.")

else:
    print("Tidak dapat memastikan izin robots.txt.")
    print("Jangan lanjut scraping sebelum diperiksa.")
```

```python

from bs4 import BeautifulSoup
import pandas as pd

# Menyiapkan tempat penyimpanan data
data_wisata = []

# Mengambil semua heading nama wisata
for h3 in soup.find_all("h3"):

    nama = h3.get_text(" ", strip=True)

    # Mengambil kategori dari heading h2 sebelumnya
    kategori_tag = h3.find_previous("h2")
    kategori = kategori_tag.get_text(" ", strip=True) if kategori_tag else "Tidak diketahui"

    # Mengambil link yang sesuai dengan nama wisata
    link_tag = h3.find_next("a")

    if link_tag and link_tag.get_text(" ", strip=True) == nama:
        link = link_tag.get("href")
    else:
        link = None

    data_wisata.append({
        "nama_wisata": nama,
        "kategori": kategori,
        "link": link
    })

# Mengubah menjadi DataFrame
df = pd.DataFrame(data_wisata)

# Menampilkan hasil
print("Jumlah data wisata:", len(df))
display(df.head(20))
```

```python

import requests
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin

# Daftar halaman sumber
halaman = {
    "Best Attractions": "https://visitbali.com/best-attractions",
    "Best Destinations": "https://visitbali.com/best-destinations",
    "Experiences": "https://visitbali.com/experiences"
}

semua_data = []

# Mengambil data dari setiap halaman
for nama_halaman, url in halaman.items():

    response = requests.get(url, timeout=30)
    response.raise_for_status()

    soup_halaman = BeautifulSoup(response.text, "html.parser")

    for h3 in soup_halaman.find_all("h3"):

        nama = h3.get_text(" ", strip=True)

        kategori_tag = h3.find_previous("h2")
        kategori = (
            kategori_tag.get_text(" ", strip=True)
            if kategori_tag else "Tidak diketahui"
        )

        link_tag = h3.find_next("a")

        if link_tag and link_tag.get_text(" ", strip=True) == nama:
            link = urljoin(url, link_tag.get("href", ""))
        else:
            link = None

        semua_data.append({
            "nama_wisata": nama,
            "kategori": kategori,
            "link": link,
            "sumber_halaman": nama_halaman
        })

# Membuat DataFrame
df_gabungan = pd.DataFrame(semua_data)


df_gabungan = df_gabungan.reset_index(drop=True)

# Menampilkan hasil
print("Total data setelah digabung:", len(df_gabungan))
print("\nJumlah data per sumber:")
print(df_gabungan["sumber_halaman"].value_counts())

display(df_gabungan.head(20))
```
