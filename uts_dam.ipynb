```python

import pandas as pd
import os
import re
import unicodedata

# Lokasi folder data
folder = "/content/proyek_uts_pariwisata/data/processed"

# Membaca dua dataset
df_tour = pd.read_csv(folder + "/tour_clean.csv")
df_scraping = pd.read_csv(folder + "/visitingjogja_yogyakarta.csv")

# Fungsi untuk menyamakan format nama
def normalisasi_nama(nama):
    nama = str(nama).lower().strip()
    nama = unicodedata.normalize("NFKD", nama)
    nama = "".join(
        karakter for karakter in nama
        if not unicodedata.combining(karakter)
    )
    nama = re.sub(r"[^a-z0-9 ]", "", nama)
    nama = re.sub(r"\s+", " ", nama)
    return nama

# Membuat kolom nama yang sudah dinormalisasi
df_tour["nama_normal"] = df_tour["Place_Name"].apply(normalisasi_nama)
df_scraping["nama_normal"] = df_scraping["judul_wisata"].apply(normalisasi_nama)

# Mencari nama tempat yang cocok
nama_cocok = set(df_tour["nama_normal"]) & set(df_scraping["nama_normal"])

print("JUMLAH DATA KAGGLE:", len(df_tour))
print("JUMLAH DATA SCRAPING:", len(df_scraping))
print("JUMLAH NAMA YANG COCOK:", len(nama_cocok))

# Menampilkan nama yang cocok
hasil_cocok = df_tour[
    df_tour["nama_normal"].isin(nama_cocok)
][["Place_Name", "Category", "City"]]

display(hasil_cocok)
```

```python

from difflib import SequenceMatcher
import pandas as pd

# Membandingkan nama tempat wisata
kandidat = []

for nama_kaggle in df_tour["Place_Name"].dropna():
    skor_terbaik = []

    for nama_scraping in df_scraping["judul_wisata"].dropna():
        skor = SequenceMatcher(
            None,
            normalisasi_nama(nama_kaggle),
            normalisasi_nama(nama_scraping)
        ).ratio() * 100

        skor_terbaik.append((nama_scraping, round(skor, 2)))

    # Mengambil 3 kandidat dengan skor tertinggi
    skor_terbaik = sorted(
        skor_terbaik,
        key=lambda x: x[1],
        reverse=True
    )[:3]

    for nama_scraping, skor in skor_terbaik:
        if skor >= 60:
            kandidat.append({
                "Nama_Kaggle": nama_kaggle,
                "Kandidat_Scraping": nama_scraping,
                "Skor_Kemiripan": skor
            })

df_kandidat = pd.DataFrame(kandidat)

df_kandidat = df_kandidat.sort_values(
    by="Skor_Kemiripan",
    ascending=False
)

print("JUMLAH KANDIDAT KECocokan:", len(df_kandidat))
display(df_kandidat)
```
